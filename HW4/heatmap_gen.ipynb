```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------------------
# Your confusion matrices
# ---------------------------------------

train_cm = np.array([
    [375, 0, 1, 2, 1, 6, 6, 2, 6, 1],
    [0, 378, 6, 1, 1, 0, 4, 1, 9, 0],
    [6, 4, 337, 8, 2, 0, 6, 17, 17, 3],
    [2, 3, 11, 330, 2, 7, 4, 19, 16, 6],
    [2, 5, 1, 0, 313, 2, 7, 6, 10, 54],
    [10, 6, 1, 23, 6, 285, 11, 15, 38, 5],
    [10, 5, 2, 1, 3, 7, 368, 1, 3, 0],
    [3, 14, 8, 2, 6, 1, 2, 341, 4, 19],
    [5, 14, 7, 12, 3, 5, 7, 4, 340, 3],
    [4, 6, 1, 6, 7, 0, 1, 26, 10, 339]
])

test_cm = np.array([
    [94, 0, 0, 0, 0, 2, 3, 0, 1, 0],
    [0, 93, 2, 1, 0, 0, 0, 1, 2, 1],
    [2, 3, 82, 3, 1, 0, 1, 4, 4, 0],
    [0, 0, 6, 81, 0, 2, 0, 4, 5, 2],
    [0, 3, 1, 0, 67, 0, 4, 1, 2, 22],
    [3, 2, 1, 3, 1, 74, 2, 3, 10, 1],
    [3, 0, 2, 0, 0, 2, 93, 0, 0, 0],
    [0, 2, 3, 2, 1, 0, 1, 86, 0, 5],
    [1, 1, 1, 3, 2, 1, 4, 4, 79, 4],
    [1, 1, 0, 1, 3, 1, 0, 7, 0, 86]
])

labels = list(range(10))


# ---------------------------------------
# Function to plot a heatmap
# ---------------------------------------

def plot_confusion_matrix(cm, title, filename=None):
    plt.figure(figsize=(8, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="viridis",    # nice scientific colormap
        xticklabels=labels,
        yticklabels=labels
    )

    plt.xlabel("Predicted Class")
    plt.ylabel("True Class")
    plt.title(title)
    plt.tight_layout()

    if filename:
        plt.savefig(filename, dpi=300)
        print(f"Saved: {filename}")

    plt.show()


# ---------------------------------------
# Generate the plots
# ---------------------------------------

plot_confusion_matrix(train_cm, "Training Confusion Matrix", "training_confusion_matrix.png")
plot_confusion_matrix(test_cm, "Test Confusion Matrix", "test_confusion_matrix.png")
```
