```python
model = openai.OpenAI(api_key=OPENAI_API_KEY)

def get_response(prompt):
    response = model.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "List the first five words that come to your mind when you hear the word X. Only give me the five words, and nothing else."},
            {"role":"user", "content": prompt}
        ]
    )

    return response.choices[0].message.content

# Part 1
def store_responses_part1(cue, response):
    file_path = r'C:\Users\User\projects\Intelligent_Systems\HW1\GPT_Responses\GPT_responses1.csv'

    with open(file_path, mode="a", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow([cue, response])
```

```python
# My cues are below
cues = {
    "car": "Drive, Park, Drift, Buy, Sell",
    "plant": "Water, Tea, Aloe Vera, Mint, Green",
    "warm": "Summer, Sun, Love, Wholesome, Sweater",
    "heavy": "Bag, Weights, Kilograms, Pounds, Emotions",
    "read": "Book, Emotions, Red, Minds, Stories",
    "coffee": "Energy, Sweet, Delicious, Milk, Anxiety",
    "yogurt": "Protein, Milk, Fermented, Sour, Spoon"
}

raw_ques = list(cues.keys())

for cue in raw_ques:
    prompt = f"List the first five words that come to your mind when you hear the word {cue}. Only give me the five words, and nothing else. No full stops either."
    for i in range(0, 10):
        response = get_response(prompt)
        store_responses_part1(cue, response)

print("Done")
```
